```python
from itertools import product

def evaluate(expr, m):
    return eval(expr, {}, m)

def entails(kb, q, symbols):
    for vals in product([False, True], repeat=len(symbols)):
        m = dict(zip(symbols, vals))
        if all(evaluate(e, m) for e in kb) and not evaluate(q, m):
            return False
    return True

kb = ["P == Q", "Q or R", "P or R"]
symbols = ["P", "Q", "R"]
queries = {"Q": "Q", "P -> R": "(not P) or R",
           "R -> Q": "(not R) or Q"}

print("P Q R | KB")
for vals in product([False, True], repeat=3):
    m = dict(zip(symbols, vals))
    k = all(evaluate(e, m) for e in kb)
    print(*map(int, vals), "|", int(k))

print("\nENTAILMENT RESULTS:")
for name, q in queries.items():
    result = entails(kb, q, symbols)
    print(f"KB |= {name}: {result}")
    if result:
        print(f"KB entails {name}")
    else:
        print(f"KB does not entail {name}")
```
