```python
# Tic Tac Toe

board = ["1", "2", "3",
         "4", "5", "6",
         "7", "8", "9"]

def display():
    print(board[0], "|", board[1], "|", board[2])
    print("--+---+--")
    print(board[3], "|", board[4], "|", board[5])
    print("--+---+--")
    print(board[6], "|", board[7], "|", board[8])


def check_win(p):
    if (board[0] == p and board[1] == p and board[2] == p) or \
       (board[3] == p and board[4] == p and board[5] == p) or \
       (board[6] == p and board[7] == p and board[8] == p) or \
       (board[0] == p and board[3] == p and board[6] == p) or \
       (board[1] == p and board[4] == p and board[7] == p) or \
       (board[2] == p and board[5] == p and board[8] == p) or \
       (board[0] == p and board[4] == p and board[8] == p) or \
       (board[2] == p and board[4] == p and board[6] == p):
        return True
    return False

print("TIC TAC TOE")
display()

for turn in range(9):

    if turn % 2 == 0:
        player = "X"
    else:
        player = "O"

    position = int(input("Player " + player + ", enter position (1-9): "))

    board[position - 1] = player

    display()

    if check_win(player):
        print(player, "wins!")
        break

else:
    print("Game is a draw!")
```

```python
# Vacuum Cleaner Agent

roomA = input("Enter Room A status (Dirty/Clean): ")
roomB = input("Enter Room B status (Dirty/Clean): ")

location = input("Enter vacuum location (A/B): ").upper()

print("\n--- Vacuum Cleaner Agent ---")

# Simple Reflex Agent
while roomA == "Dirty" or roomB == "Dirty":

    if location == "A":

        if roomA == "Dirty":
            print("Room A is Dirty -> Suck")
            roomA = "Clean"
        else:
            print("Room A is Clean -> Move Right")
            location = "B"

    else:

        if roomB == "Dirty":
            print("Room B is Dirty -> Suck")
            roomB = "Clean"
        else:
            print("Room B is Clean -> Move Left")
            location = "A"

print("\nBoth rooms are Clean!")
print("Goal Achieved!")
```
