```python

# First Order Logic - Forward Chaining

# Initial Knowledge Base
facts = {
    ("American", "Robert"),
    ("Enemy", "A", "America"),
    ("Owns", "A", "T1"),
    ("Missile", "T1")
}

# Forward Chaining
def forward_chaining(facts):

    new_facts = True

    while new_facts:
        new_facts = False

        # Rule 1: Enemy of America is hostile
        for fact in list(facts):
            if fact[0] == "Enemy":
                x = fact[1]
                result = ("Hostile", x)

                if result not in facts:
                    facts.add(result)
                    print("Derived:", result)
                    new_facts = True

        # Rule 2: Missiles are weapons
        for fact in list(facts):
            if fact[0] == "Missile":
                x = fact[1]
                result = ("Weapon", x)

                if result not in facts:
                    facts.add(result)
                    print("Derived:", result)
                    new_facts = True

        # Rule 3: All missiles owned by A
        # were sold by Robert to A
        if (("Missile", "T1") in facts and
            ("Owns", "A", "T1") in facts):

            result = ("Sells", "Robert", "T1", "A")

            if result not in facts:
                facts.add(result)
                print("Derived:", result)
                new_facts = True

        # Rule 4: Crime rule
        if (("American", "Robert") in facts and
            ("Weapon", "T1") in facts and
            ("Sells", "Robert", "T1", "A") in facts and
            ("Hostile", "A") in facts):

            result = ("Criminal", "Robert")

            if result not in facts:
                facts.add(result)
                print("Derived:", result)
                new_facts = True

    return facts


# Execute Forward Chaining
result = forward_chaining(facts)

# Check the Query
query = ("Criminal", "Robert")

if query in result:
    print("\nQuery Proved: Robert is Criminal")
else:
    print("\nQuery Not Proved")
```
