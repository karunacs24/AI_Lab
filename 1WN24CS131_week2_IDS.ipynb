```python
# 8 Puzzle using DFS

def dfs(start, goal):
    stack = [(start, [])]

    while stack:
        state, path = stack.pop()

        # Goal reached
        if state == goal:
            return path + [state]

        # Stop after 6 moves
        if len(path) >= 6:
            continue

        # Find blank tile
        zero = state.index(0)

        row = zero // 3
        col = zero % 3

        moves = []

        # LEFT
        if col > 0:
            moves.append(zero - 1)

        # UP
        if row > 0:
            moves.append(zero - 3)

        # DOWN
        if row < 2:
            moves.append(zero + 3)

        # RIGHT
        if col < 2:
            moves.append(zero + 1)

        # Generate new states
        for move in reversed(moves):

            new_state = list(state)

            # Move blank tile
            new_state[zero], new_state[move] = \
                new_state[move], new_state[zero]

            new_state = tuple(new_state)

            stack.append((new_state, path + [state]))

    return None


# Take input
print("Enter 9 numbers (0 represents blank):")
start = tuple(map(int, input().split()))

# Goal state
goal = (
    1, 2, 3,
    4, 5, 6,
    7, 0, 8
)

# Apply DFS
solution = dfs(start, goal)

# Display result
if solution:

    print("\nSolution found!")
    print("Number of moves:", len(solution) - 1)

    for state in solution:
        print()
        print(state[0], state[1], state[2])
        print(state[3], state[4], state[5])
        print(state[6], state[7], state[8])

else:
    print("No solution found within 6 moves.")
```

```python
def DLS(graph, node, goal, depth):
    if node == goal:
        return True

    if depth == 0:
        return False

    for child in graph[node]:
        if DLS(graph, child, goal, depth - 1):
            return True

    return False


def IDDFS(graph, start, goal):
    depth = 0

    while True:
        if DLS(graph, start, goal, depth):
            return True
        depth += 1


# Take graph input
graph = {}

n = int(input("Enter number of nodes: "))

for i in range(n):
    node = input("Enter node: ")
    children = input("Enter children separated by space: ").split()
    graph[node] = children

# Take start and goal
start = input("Enter start node: ")
goal = input("Enter goal node: ")

# IDDFS
if IDDFS(graph, start, goal):
    print("Goal found!")
else:
    print("Goal not found!")
```
