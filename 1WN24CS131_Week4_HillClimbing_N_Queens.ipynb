```python

import random

N = 4

def cost(b):
    return sum(
        1 for i in range(N) for j in range(i + 1, N)
        if b[i] == b[j] or abs(b[i] - b[j]) == abs(i - j)
    )

def solve(b):
    while True:
        print("Current:", [x + 1 for x in b], "Cost:", cost(b))

        if cost(b) == 0:
            return b

        neighbors = []
        for c in range(N):
            for r in range(N):
                if r != b[c]:
                    t = b.copy()
                    t[c] = r
                    neighbors.append(t)

        best_cost = min(cost(t) for t in neighbors)

        if best_cost >= cost(b):
            print("Stuck! Restarting with a random board.")
            b = [random.randrange(N) for _ in range(N)]
        else:
            b = random.choice([t for t in neighbors
                               if cost(t) == best_cost])

b = list(map(int, input("Enter 4 rows (1-4): ").split()))

if len(b) != N or any(x < 1 or x > N for x in b):
    print("Enter exactly 4 numbers between 1 and 4.")
else:
    answer = solve([x - 1 for x in b])
    print("Solution:", [x + 1 for x in answer])
    print("Final cost:", cost(answer))
```
